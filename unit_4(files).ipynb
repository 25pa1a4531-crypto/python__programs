file handling in python

create a file

```python
file=open("new.txt","w")
file.write("apple")
print("file created")
file.close()
```

read a file

```python
file= open("new.txt","r")
data=file.read()
print(data)
file.close()
```

update a file

```python
with open ("new.txt","a")as f:
  f.write("\nnew line")
  print("file update")
```

program to sort words in a file                                                                              

```python
with open("new.txt","r") as f:
  data=f.read().split()
  data_lower=[i.lower() for i in data]
  sort_data=sorted(data_lower)
  with open("f1.txt","w") as f2:
    for i in sort_data:
      print("done")
```

copy the contents of one file to another file

```python
with open("new.txt","r") as f:
    data=f.readline()
with open("f1.txt","a") as f2:
      f2.write(data)
```

```python

#Print reverse of the number using functions
def reverse_num(n):
 rev=0
while (n!=0):
  digit=n%10
  rev=rev*10+digit
  n=n//10
return rev
num=int(input("Enter a number:"))
print("reverse number:",reverse_num(num))
```
